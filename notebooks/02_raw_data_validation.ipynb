```python
import sys
import json
from pathlib import Path


PROJECT_ROOT = Path.cwd().parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))
```

```python
from src.data.validation import validate_episode
```

```python
DATA_DIR = PROJECT_ROOT / "Datasets"

json_files = sorted(DATA_DIR.glob("*.json"))

print("Total JSON files:", len(json_files))
```

```python
sample_file = json_files[0]

with open(sample_file, "r", encoding="utf-8") as f:
    first_data = json.load(f)

validate_episode(first_data)

print("Sample JSON validation passed ✅")
```

```python
validated = 0
failed = []

for i, file_path in enumerate(json_files, start=1):

    try:
        with open(file_path, "r", encoding="utf-8") as f:
            data = json.load(f)

        validate_episode(data)
        validated += 1

    except Exception as error:
        failed.append({
            "file": file_path.name,
            "error": str(error)
        })

    if i % 50 == 0 or i == len(json_files):
        print(f"Validated {i}/{len(json_files)}")
```

```python
print("Total files:", len(json_files))
print("Valid files:", validated)
print("Failed files:", len(failed))
```

```python
if failed:
    print("\nFailed files:")
    for item in failed:
        print(item)
else:
    print("\nAll raw JSON files passed validation ✅")
```
