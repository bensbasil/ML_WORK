```python
from pathlib import Path
import json
import pandas as pd
```

```python
DATA_DIR = Path("../Datasets")

json_files = sorted(DATA_DIR.glob("*.json"))

print(f"Total JSON files: {len(json_files)}")
print("First file:", json_files[0].name)
```

```python
sample_file = json_files[0]

with open(sample_file, "r", encoding="utf-8") as f:
    first_data = json.load(f)

print("Loaded:", sample_file.name)
```

```python
for key, value in first_data.items():
    print(f"{key}: {type(value).__name__}")
```

```python
for key in ["configuration", "info", "rewards", "statuses", "steps"]:
    value = first_data[key]

    print(f"\n--- {key} ---")
    print("Type:", type(value).__name__)

    if isinstance(value, dict):
        print("Keys:", value.keys())

    elif isinstance(value, list):
        print("Length:", len(value))

        if value:
            print("First item type:", type(value[0]).__name__)
```

```python
step = first_data["steps"][0]

print("Step type:", type(step).__name__)
print("Number of players:", len(step))

for player, agent in enumerate(step):
    print(f"\nPlayer {player}")
    print("Keys:", agent.keys())
```

```python
agent = first_data["steps"][0][0]

print("Action:")
print(agent["action"])

print("\nReward:")
print(agent["reward"])

print("\nStatus:")
print(agent["status"])

print("\nObservation keys:")
print(agent["observation"].keys())
```

```python
obs = agent["observation"]

for key, value in obs.items():
    print(f"{key}: {type(value).__name__}")
```

```python
player = 0

farm = (
    first_data["steps"][0][player]
    ["observation"]["farms"][player]
)

for key, value in farm.items():
    print(f"{key}: {type(value).__name__}")
```

```python
step_number = 100
player = 0

money = first_data["steps"][step_number][player]["observation"]["farms"][player]["money"]

print(money)
```

```python
prices = (
    first_data["steps"][0][0]
    ["observation"]["market"]["prices"]
)

print(prices)
```

```python
step_number = 100
player = 0

carrot_price = (
    first_data["steps"][step_number][player]
    ["observation"]["market"]["prices"]["CARROT"]
)

print(carrot_price)
```

```python
private = (
    first_data["steps"][0][0]
    ["observation"]["private"]
)

print("SEEDS")
print(private["seeds"])

print("\nSHED")
print(private["shed"])
```

```python
for step_number in [0, 100, 300, 719]:

    agent = first_data["steps"][step_number][0]
    farm = agent["observation"]["farms"][0]

    print(f"\n===== STEP {step_number} =====")
    print("Money:", farm["money"])
    print("Farmer:", farm["farmer"])
```

```python
unique_tile_examples = {}

for step_data in first_data["steps"]:
    for player, agent in enumerate(step_data):

        farm = agent["observation"]["farms"][player]

        for row in farm["tiles"]:
            for tile in row:

                if isinstance(tile, dict):

                    tile_type = tile.get("kind", "UNKNOWN")

                    if tile_type not in unique_tile_examples:
                        unique_tile_examples[tile_type] = tile


for tile_type, example in unique_tile_examples.items():

    print(f"\n--- {tile_type} ---")
    print("Example:", example)
    print("Keys:", example.keys())
```
